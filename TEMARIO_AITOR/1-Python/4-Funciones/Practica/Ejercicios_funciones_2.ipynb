````python
def contador(text,letra):
    

    veces=text.count(letra.upper())+text.count(letra.lower())
    return text,letra,veces

contador("Hola que hase Amigo","a")
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci (n):

    if n >= 0:
        if n== 0 or n ==1:
         return n

        else:
         return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(9))        
````
