# Ex - GroupBy

```python
import pandas as pd
df= pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv")
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

### Step 3. Assign it to a variable called drinks.

```python
drinks = df
```

### Step 4. Which continent drinks more beer on average?

```python
print(df.columns)
beer= drinks.groupby("continent").agg(
    beer_average= ("beer_servings","mean")
).sort_values(by="beer_average", ascending=False)
print(beer)
```

### Step 5. For each continent print the statistics for wine consumption.

```python
wine= drinks.groupby("continent").agg(
    wine_consumption=("wine_servings","sum")
).sort_values(by="wine_consumption", ascending=False)
print(wine)
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
alcohol= drinks.groupby("continent").agg(
   beer_consumption=("beer_servings","mean"),
   spirit_consumption=("spirit_servings","mean"),
   wine_consumption=("wine_servings","mean"),
   total_consumption=("total_litres_of_pure_alcohol","mean")
).sort_values(by="total_consumption", ascending=False)
print(alcohol)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
alcohol_median= drinks.groupby("continent").agg(
   beer_consumption=("beer_servings","median"),
   spirit_consumption=("spirit_servings","median"),
   wine_consumption=("wine_servings","median"),
   total_consumption=("total_litres_of_pure_alcohol","median"),
).sort_values(by="total_consumption", ascending=False)
print(alcohol_median)
```

### Step 8. Print the mean, min and max values for spirit consumption per continent.

```python
spirit_consumption_metrics= df.groupby("continent").agg(
    spirit_consumption_mean=("spirit_servings", "mean"),
    spirit_consumption_min=("spirit_servings", "min"),
    spirit_consumption_max=("spirit_servings", "max")
).sort_values(by="continent", ascending=False)
print(spirit_consumption_metrics)
```
