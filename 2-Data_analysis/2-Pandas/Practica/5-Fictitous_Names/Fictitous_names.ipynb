# Fictitious Names

### Step 2. Create the 3 DataFrames based on the following raw data

```python
raw_data_1 = {
        'subject_id': ['1', '2', '3', '4', '5'],
        'first_name': ['Alex', 'Amy', 'Allen', 'Alice', 'Ayoung'],
        'last_name': ['Anderson', 'Ackerman', 'Ali', 'Aoni', 'Atiches']}
raw_data_2 = {
        'subject_id': ['4', '5', '6', '7', '8'],
        'first_name': ['Billy', 'Brian', 'Bran', 'Bryce', 'Betty'],
        'last_name': ['Bonder', 'Black', 'Balwner', 'Brice', 'Btisan']}
raw_data_3 = {
        'subject_id': ['1', '2', '3', '4', '5', '7', '8', '9', '10', '11'],
        'test_id': [51, 15, 15, 61, 16, 14, 15, 1, 61, 16]}

```

### Step 3. Assign each to a variable called data1, data2, data3

```python
import pandas as pd

data1 = pd.DataFrame(raw_data_1)
data2 = pd.DataFrame(raw_data_2)
data3 = pd.DataFrame(raw_data_3)
display(data1,data2,data3)
```

### Step 4. Join the two first dataframes along rows and assign to all_data

```python

all_data2 = pd.concat([data1,data2], ignore_index=True)
print(all_data2)
print(all_data2.shape)
```

### Step 5. Join the two first dataframes along columns and assing to all_data_col

```python
print(data1.columns)
print(data2.columns)
all_data = data1.merge(
    data2,
    on=["first_name", "last_name","subject_id"],
    how= "outer"
).sort_values(by = "subject_id", ascending = True)
print(all_data)
all_data2 = all_data.drop(["subject_id_x","subject_id_y"], axis=1)
print(all_data2)
print(all_data2.shape)
```

### Step 6. Print the unique values of data3

```python
print((data3["test_id"]).value_counts())
print((data3["test_id"].nunique()))
print((data3["test_id"].unique()))
```

### Step 7. Merge all_data and data3 along the subject_id value

```python
final = all_data2.merge(
    data3,
    on="subject_id",
    how="left"
)
final = final.drop(["subject_id"], axis=1)
print(final)
print(final.shape)
```

### Step 8. Merge only the data that has the same 'subject_id' on both data1 and data2

```python
final2= data1.merge(
    data2,
    on="subject_id",
    how= "inner"
)
print(final2)
first_name1 = final2["first_name_x"]
first_name2= final2["first_name_y"]
print(first_name1)
print(first_name2)
final3 = first_name1.merge(
    first_name2,
    left_on = "first_name_x",
    right_on= "first_name_y",
    how="outer"
)
print(final3)
```
