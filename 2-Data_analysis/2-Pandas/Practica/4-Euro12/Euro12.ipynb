### Step 7. View only the columns Team, Yellow Cards and Red Cards and assign them to a dataframe called discipline

```python
discipline = euro12[["Team","Yellow Cards","Red Cards"]]
print(discipline)
```

### Step 8. Sort the teams by Red Cards, then to Yellow Cards

```python
print(discipline.sort_values(by="Yellow Cards", ascending = False))
print("---------------------------------------------------")
print(discipline.sort_values(by="Red Cards", ascending = False))
print()
```

### Step 9. Calculate the mean Yellow Cards given per Team

```python
yellow_cards= euro12.groupby("Team").agg(
    yellow_cards_mean = ("Yellow Cards","mean")
).sort_values(by = "yellow_cards_mean", ascending = False)
print(yellow_cards)
```
